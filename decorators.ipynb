```python
# %%
import random
from functools import wraps


def repeat(iterations: int = 10):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            iterations_data = []
            for it in range(iterations):
                result = func(*args, **kwargs)
                iterations_data.append(result)
            iterations_data = sorted(iterations_data, reverse=True)
            print(iterations_data)

        return wrapper
    return decorator


@repeat(iterations=20)
def lotto_draw(start, end):
    return random.randint(start, end)


lotto_draw(1, 49)
```

```python
from functools import wraps
import time
from uuid import uuid4

# Users
user1 = {"id":uuid4(),"attempts":0}
user2 = {"id":uuid4(),"attempts":0}

def download_limiter(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        user = args[0] if args else kwargs["user"]
        attempts = user.get("attempts")
        
        if attempts > 0: 
            print(f"Attempt no. {attempts + 1}. Waiting {attempts} secs...")
        user["attempts"] = max(1, attempts * 2)
        time.sleep(attempts)
        func(*args, **kwargs)
    return wrapper

@download_limiter
def download(user):
    print(f"User {user['id']}, your download is ready.")
    
download(user = user1)
download(user1)
download(user1)
download(user2)
download(user2)
download(user1)
download(user2)
```
